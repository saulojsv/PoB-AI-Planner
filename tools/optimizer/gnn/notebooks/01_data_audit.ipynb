# PoB AI Planner — auditoria de shards

Este notebook só admite observações recalculadas pelo PoB, com direitos/consentimento revisados e sem mistura de engine, skill, cenário ou linhagens entre partições. O modo padrão é somente auditoria; a admissão para `02_Validated_Datasets` é uma ação separada e explícita.

```python
from google.colab import drive
from pathlib import Path
import zipfile, sys, subprocess
drive.mount('/content/drive')
ROOT = Path('/content/drive/MyDrive/00 - Projetos/PoB - IA')
SOURCE_ZIP = ROOT / '05_Notebooks/pob_gnn_source.zip'
assert SOURCE_ZIP.is_file(), f'Pacote de código ausente: {SOURCE_ZIP}'
SOURCE = Path('/content/pob_ai_source')
SOURCE.mkdir(exist_ok=True)
with zipfile.ZipFile(SOURCE_ZIP) as archive: archive.extractall(SOURCE)
sys.path.insert(0, str(SOURCE))
INBOX = ROOT / '01_Inbox'
VALIDATED = ROOT / '02_Validated_Datasets'
REPORT = ROOT / '07_Reports/dataset_audit.json'
subprocess.run([sys.executable, '-m', 'tools.optimizer.gnn.audit_shards', '--inbox', str(INBOX), '--validated-root', str(VALIDATED), '--report', str(REPORT)], cwd=SOURCE, check=True)
```
