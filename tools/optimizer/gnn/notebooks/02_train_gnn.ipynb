# PoB AI Planner — treino do ensemble HeteroGATv2

O treino lê exclusivamente `02_Validated_Datasets/train` e `validation`. O código não abre `holdout`. Os pesos são `state_dict` tensor-only, e o checkpoint fica em `03_Models_Staging`. Não há promoção automática.

```python
from google.colab import drive
from pathlib import Path
import zipfile, sys, subprocess
drive.mount('/content/drive')
ROOT = Path('/content/drive/MyDrive/00 - Projetos/PoB - IA')
SOURCE_ZIP = ROOT / '05_Notebooks/pob_gnn_source.zip'
SOURCE = Path('/content/pob_ai_source'); SOURCE.mkdir(exist_ok=True)
with zipfile.ZipFile(SOURCE_ZIP) as archive: archive.extractall(SOURCE)
sys.path.insert(0, str(SOURCE))
subprocess.run([sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check', 'torch==2.6.0', '--index-url', 'https://download.pytorch.org/whl/cpu'], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '--disable-pip-version-check', 'torch-geometric==2.6.1'], check=True)
DATA = ROOT / '02_Validated_Datasets'
STAGING = ROOT / '03_Models_Staging'
subprocess.run([sys.executable, '-m', 'tools.optimizer.gnn.train', '--data-root', str(DATA), '--staging', str(STAGING), '--epochs', '30', '--ensemble-members', '5', '--hidden-size', '64'], cwd=SOURCE, check=True)
```
